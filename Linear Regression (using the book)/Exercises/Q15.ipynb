```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ISLP
from ISLP import load_data
from ISLP.models import ModelSpec as MS, summarize
import statsmodels.api as sm
```

```python
boston = load_data("boston")
```

```python
boston.head(5)
```

```python
y= boston["crim"]
X = boston.drop(columns='crim')
X.insert(0,'constant',1)
```

```python
X.head()
```

```python
y.head()
```

```python
X.columns
```

```python
len(X.columns)
```

```python
model_1 = sm.OLS(y,X[['constant','zn']]).fit()
model_2 = sm.OLS(y,X[['constant','indus']]).fit()
model_3 = sm.OLS(y,X[['constant','chas']]).fit()
model_4 = sm.OLS(y,X[['constant','nox']]).fit()
model_5 = sm.OLS(y,X[['constant','rm']]).fit()
model_6 = sm.OLS(y,X[['constant','age']]).fit()
model_7 = sm.OLS(y,X[['constant','dis']]).fit()
model_8 = sm.OLS(y,X[['constant','rad']]).fit()
model_9 = sm.OLS(y,X[['constant','tax']]).fit()
model_10 = sm.OLS(y,X[['constant','ptratio']]).fit()
model_11 = sm.OLS(y,X[['constant','lstat']]).fit()
model_12 = sm.OLS(y,X[['constant','medv']]).fit()
```
