```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from ISLP import load_data
import seaborn as sns
import statsmodels.api as sm
from ISLP.models import summarize
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
```

```python
rng = np.random.default_rng(1)
x = rng.normal(size=100)
y = 2 * x + rng.normal(size=100)
```

## Regressing X onto y (without Intercept)

```python
from ISLP.models import ModelSpec as MS
x = pd.DataFrame({'X':x})
y = pd.DataFrame({'Y':y})
design = MS('X', intercept=False)
X = design.fit_transform(x)
model_X_onto_y = sm.OLS(y,X).fit()
summarize(model_X_onto_y)
```

## Regressing y onto X (without Intercept)

```python
design1 = MS(['Y'], intercept=False)
y = design1.fit_transform(y)
model_y_onto_X = sm.OLS(X,y).fit()
summarize(model_y_onto_X)
```

```python
num=0;den=0
for i in range(0,len(X)):
    num= num + (X['X'].iloc[i]*y['Y'].iloc[i])
    den= den + np.square(X['X'].iloc[i])

beta_cap_of_Y_onto_X = num/den
beta_cap_of_Y_onto_X
```

```python
num1=0; den1=0
for i in range(0,len(X)):
    num1 = num1 + np.square((y['Y'].iloc[i] - (X['X'].iloc[i]*beta_cap_of_Y_onto_X)))
    den1 = den1 + np.square(X['X'].iloc[i])
den1= (len(X)-1)*den1

std_error_of_beta_cap_of_Y_onto_X = np.sqrt(num1/den1)
std_error_of_beta_cap_of_Y_onto_X
```

```python
t_stats_Y_onto_X = beta_cap_of_Y_onto_X/std_error_of_beta_cap_of_Y_onto_X
t_stats_Y_onto_X
```

```python
design2 = MS('X')
X_dummy = design2.fit_transform(x)
model_X_onto_y_using_intercept = sm.OLS(y,X_dummy).fit()
summarize(model_X_onto_y_using_intercept)
```

```python
design3 = MS(['Y'])
y_dummy = design3.fit_transform(y)
model_y_onto_X_using_intercept = sm.OLS(X,y_dummy).fit()
summarize(model_y_onto_X_using_intercept)
```
