```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ISLP
from ISLP.models import ModelSpec as MS, summarize
import statsmodels.api as sm
from narwhals import DataFrame
```

(a). The regression coefficient of Y onto X and X onto Y will be equal when Y = X

(c)

```python
np.random.seed(42)
X = np.random.rand(100)
Y = X
```

```python
plt.scatter(X,Y)
```

```python
X = pd.DataFrame({"X":X})
Y = pd.DataFrame({"Y":Y})
```

```python
design = MS('X', intercept=False)
X_dummy = design.fit_transform(X)
model_X_onto_y = sm.OLS(Y,X_dummy).fit()
summarize(model_X_onto_y)
```

```python
design1 = MS(['Y'], intercept=False)
y_dummy = design1.fit_transform(Y)
model_y_onto_X = sm.OLS(X,y_dummy).fit()
summarize(model_y_onto_X)
```
