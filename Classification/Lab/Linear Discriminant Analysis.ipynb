## Applying LDA on all the features

```python
X= Smarket.drop(columns= ['Year', 'Today', 'Direction'])
y= Smarket.Direction == 'Up'
```

```python
train= Smarket.Year<2005
X_train, X_test= X.loc[train], X.loc[~train]
y_train, y_test= y.loc[train], y.loc[~train]
```

```python
lda= LDA(store_covariance=True)
lda.fit(X_train, y_train)
```

```python
lda.means_
```

```python
lda.classes_
```

```python
lda.priors_
# Prior Probability
```

```python
lda.scalings_
# It contains the coefficients (weights) of the linear discriminants, representing the directions that maximize between-class variance while minimizing within-class variance.
```

```python
lda_pred= lda.predict(X_test)
```

```python
L_test= Smarket.Direction.loc[~train]
confusion_table(lda_pred, y_test)
```

## Applying LDA on the least P-values features

```python
X_lp= Smarket[['Lag1', 'Lag2']]

X_lp_train= X_lp.loc[train]
X_lp_test= X_lp.loc[~train]
```

```python
lda_lp= LDA(store_covariance=True)
lda_lp.fit(X_lp_train, y_train)
```

```python
lda_pred_lp= lda_lp.predict(X_lp_test)
```

```python
lda_pred_lp= np.where(lda_pred_lp == True, 'Up', 'Down')
```

```python
y_test_lp= y_test.map({True: 'Up', False: 'Down'})
```

```python
confusion_table(lda_pred_lp, y_test_lp)
```

```python
lda_lp.priors_
```

```python
lda_lp.scalings_
```

```python
lda_lp.means_
```

```python
lda_lp.classes_
```

```python
lda_prob_lp= lda_lp.predict_proba(X_lp_test)
np.all(np.where(lda_prob_lp[:,1] >= 0.5, 'Up', 'Down') == lda_pred_lp)
```

```python
np.sum(lda_prob_lp[:,0] > 0.9)
```

```python
np.mean(lda_pred_lp==y_test_lp)
```
