```python
import numpy as np
import pandas as pd
from matplotlib.pyplot import subplots
import statsmodels.api as sm
from ISLP import load_data
from ISLP.models import (ModelSpec as MS,
                         summarize)
from ISLP import confusion_table
from statsmodels.genmod.families import Binomial
```

```python
Smarket= load_data('Smarket')
Smarket.head()
```

```python
Smarket.columns
```

```python
Smarket.drop(columns='Direction').corr()
```

```python
Smarket.plot(y='Volume') # the average no. of shares traded daily increased from 2001 to 2005
```

```python
allvars= Smarket.columns.drop(['Today', 'Direction', 'Year'])

design= MS(allvars)
X= design.fit_transform(Smarket) # adds the intercept in the Days Dataframe
y= Smarket.Direction == 'Up' # True, False Series

glm = sm.GLM(y, X, family=sm.families.Binomial())

""" sm.families.Binomial() -> tells the model that the data
follows a binomial distribution, which defaults to the
logit link function for standard logistic regression """

results= glm.fit()
summarize(results)
```

```python
results.params
```

```python
results.pvalues
```

```python
probs = results.predict()
probs[:10]
```

```python
labels= np.array(['Down']*1250)
labels[probs>0.5]='Up'

confusion_table(labels, Smarket.Direction)
# comparing the predicted vs original value
```

## Splitting the data into train or test data

```python
train= (Smarket.Year<2005)
Smarket_train = Smarket.loc[train]
Smarket_test = Smarket.loc[~train]
Smarket_test.shape
```

```python
X_train, X_test = X.loc[train], X.loc[~train]
y_train, y_test = y.loc[train], y.loc[~train]

glm_train = sm.GLM(y_train, X_train, family=sm.families.Binomial())
results= glm_train.fit()
probs= results.predict(exog=X_test) # predicting on the test dataset
```
